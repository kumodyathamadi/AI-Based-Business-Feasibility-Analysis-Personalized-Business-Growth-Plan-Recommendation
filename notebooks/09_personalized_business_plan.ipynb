## Executive Summary & Research Context
A decision-support system must convert analytical model outputs (ML feasibility predictions, SHAP attribution drivers, TOPSIS strategy rankings, and What-If counterfactual boundaries) into an actionable, tailored business plan.

This notebook implements **Phase 9: Constraint-Aware Personalized Business Plan & Structured Business Profile Export**.

### Key Objectives:
1. **Multi-Source Analytical Synthesis**: Integrate outputs from Notebooks 01–08 into a cohesive, structured business/growth recommendation plan.
2. **5-Section Personalized Business/Growth Plan**:
   - **Section 1: Business Overview & Value Proposition** (Category, Stage, Target Customer Context in Sri Lanka)
   - **Section 2: Operations & Resource Allocation Plan** (Location, Staffing, Equipment, Suppliers)
   - **Section 3: Marketing & Sales Execution Plan** (Pricing, Customer Reach, Demand Mitigation)
   - **Section 4: Financial Planning & SHAP Risk Mitigation** (Budget, Capital Utilization, Risk Countermeasures)
   - **Section 5: Phased Action Roadmap** (Short-term 0-3m, Medium-term 3-12m, Long-term 1y+)
3. **Structured Business Profile Export**: Generate a clean, machine-readable JSON profile (`models/feasibility_model/structured_business_profile.json`) for future consumption by:
   - **Component 2**: Legal, Regulatory & Ethical Guidance
   - **Component 3**: Financial Monitoring & Anomaly Detection
   - **Component 4**: GNN Risk Prediction & Scalable Growth


```python
import os
import json
import numpy as np
import pandas as pd

os.makedirs("../models/feasibility_model", exist_ok=True)
os.makedirs("../reports", exist_ok=True)

print("Setup completed successfully.")
```

## 1. Load Analytical Artifacts from Prior Phases
Loading exported metadata artifacts:
- `shap_metadata.json` (Phase 5)
- `strategy_matrix_sample.json` (Phase 6)

- `what_if_analysis_sample.json` (Phase 8)


```python
# Load TOPSIS Ranked Strategies
with open("../models/feasibility_model/topsis_ranked_strategies.json", "r", encoding="utf-8") as f:
    topsis_payload = json.load(f)

# Load What-If & Counterfactual Results
with open("../models/feasibility_model/what_if_analysis_sample.json", "r", encoding="utf-8") as f:
    what_if_payload = json.load(f)

# Load SHAP Metadata
with open("../models/feasibility_model/shap_metadata.json", "r", encoding="utf-8") as f:
    shap_payload = json.load(f)

print("All prior analytical artifacts loaded successfully.")
print(f"Target Case ID: {topsis_payload['case_id']}")
print(f"Top Ranked Strategy (TOPSIS Rank 1): {topsis_payload['top_ranked_strategy']['title']}")
```

## 2. Personalized Business & Growth Plan Generator Engine
Building `PersonalizedPlanGenerator` class to synthesize inputs into structured human-readable text blocks and operational action roadmaps.


```python
class PersonalizedPlanGenerator:
    def __init__(self):
        pass
        
    def generate_plan(self, input_profile, prediction_info, top_strategy, what_if_info):
        stage = input_profile.get("business_stage", "New")
        category = input_profile.get("business_category", "SME")
        location = input_profile.get("location_type", "Colombo")
        capital = float(input_profile.get("available_capital_lkr", 500000))
        loan = float(input_profile.get("loan_amount_lkr", 0))
        budget = float(input_profile.get("monthly_budget_lkr", 150000))
        price = float(input_profile.get("expected_price_lkr", 500))
        customers = float(input_profile.get("expected_customers_per_day", 30))
        experience = float(input_profile.get("entrepreneur_experience_years", 0))
        
        # 1. Overview & Concept
        overview = {
            "business_title": f"{stage} {category} Proposal ({location})",
            "business_stage": stage,
            "business_category": category,
            "target_location": f"{location}, Colombo District, Western Province",
            "value_proposition": f"Provide high-quality {category.lower()} products tailored to local consumer demand in {location}, leveraging a {top_strategy['strategy_type'].lower()} operational model.",
            "target_customer_profile": f"Local residents, commuters, and businesses within the {location} commercial catchment area."
        }
        
        # 2. Operations & Resource Plan
        operations = {
            "operating_model": top_strategy["strategy_type"],
            "recommended_location_strategy": f"Utilize {location} location with a suitability score of {input_profile.get('location_suitability_score', 3)}/5.",
            "staffing_allocation": {
                "available_staff": int(input_profile.get("available_staff_count", 0)),
                "required_staff": int(input_profile.get("required_staff_count", 1)),
                "recommendation": "Recruit 1 additional operational staff member" if input_profile.get("required_staff_count", 1) > input_profile.get("available_staff_count", 0) else "Maintain current staffing headcount."
            },
            "equipment_readiness": f"Equipment Score: {input_profile.get('available_equipment_score', 1)}/5 (Required: {input_profile.get('required_equipment_score', 1)}/5).",
            "supplier_sourcing": f"Supplier Availability Score: {input_profile.get('supplier_availability_score', 4)}/5 (Strong local supplier connectivity in Western Province)."
        }
        
        # 3. Marketing & Customer Acquisition
        marketing = {
            "projected_daily_customers": int(customers),
            "expected_price_point_lkr": float(price),
            "customer_demand_score": float(input_profile.get("customer_demand_score", 50)),
            "competition_level": str(input_profile.get("competition_level", "Medium")),
            "customer_acquisition_strategy": "Direct localized promotions, WhatsApp customer ordering group, and digital signage" if "Digital" in top_strategy["strategy_type"] else "Direct storefront signage, footfall discounts, and local commercial B2B flyers."
        }
        
        # 4. Financial Plan & Risk Mitigation
        financials = {
            "initial_capital_available_lkr": capital,
            "requested_loan_amount_lkr": loan,
            "recommended_strategy_investment_lkr": top_strategy["initial_investment_lkr"],
            "recommended_monthly_operating_cost_lkr": top_strategy["monthly_operating_cost_lkr"],
            "expected_monthly_revenue_lkr": top_strategy["expected_monthly_revenue_lkr"],
            "projected_net_monthly_margin_lkr": top_strategy["expected_monthly_revenue_lkr"] - top_strategy["monthly_operating_cost_lkr"],
            "primary_risk_mitigation": top_strategy["shap_risk_addressed"]
        }
        
        # 5. Phased Action Roadmap
        roadmap = {
            "phase_1_short_term_0_to_3_months": [
                f"Secure operating site/setup in {location} and finalize initial inventory worth LKR {input_profile.get('initial_inventory_cost_lkr', 50000):,}.",
                f"Allocate initial setup capital of LKR {top_strategy['initial_investment_lkr']:,} into core equipment and inventory.",
                "Establish local supplier supply contracts in Colombo District."
            ],
            "phase_2_medium_term_3_to_12_months": [
                f"Scale daily customer volume toward target benchmark of {int(customers)} customers/day.",
                f"Maintain monthly operating budget within strict LKR {top_strategy['monthly_operating_cost_lkr']:,} ceiling.",
                "Evaluate initial 6-month revenue performance against target monthly projection."
            ],
            "phase_3_long_term_1_year_plus": [
                "Reinvest net monthly margins into product line diversification or secondary satellite channels.",
                "Assess expansion into neighboring sub-districts in Western Province."
            ]
        }
        
        return {
            "overview": overview,
            "operations": operations,
            "marketing": marketing,
            "financials": financials,
            "roadmap": roadmap
        }

plan_generator = PersonalizedPlanGenerator()
print("PersonalizedPlanGenerator engine initialized successfully.")
```

## 3. Generate Personalized Business Plan for Target SME
Generating full 5-section personalized business plan combining ML prediction, SHAP risk attributions, TOPSIS rank 1 strategy, and counterfactual boundaries.


```python
raw_input = topsis_payload["input_profile"]
pred_info = topsis_payload["prediction"]
top_strat = topsis_payload["top_ranked_strategy"]

generated_plan = plan_generator.generate_plan(raw_input, pred_info, top_strat, what_if_payload)

print("==================================================")
print("PERSONALIZED BUSINESS PLAN: " + generated_plan['overview']['business_title'].upper())
print("Recommended Strategy: " + top_strat['title'] + " (TOPSIS Rank 1 Score: " + f"{top_strat['TOPSIS_Score']:.4f}" + ")")
print("--------------------------------------------------")
print("")
print("[SECTION 1: BUSINESS OVERVIEW]")
print("  Concept: " + generated_plan['overview']['value_proposition'])
print("  Target Catchment: " + generated_plan['overview']['target_location'])
print("")
print("[SECTION 2: OPERATIONS & RESOURCES]")
print("  Operating Model: " + generated_plan['operations']['operating_model'])
print("  Staffing: " + generated_plan['operations']['staffing_allocation']['recommendation'])
print("  Equipment: " + generated_plan['operations']['equipment_readiness'])
print("")
print("[SECTION 3: MARKETING & SALES]")
print("  Target Daily Customers: " + str(generated_plan['marketing']['projected_daily_customers']) + " customers/day @ LKR " + f"{generated_plan['marketing']['expected_price_point_lkr']:,.2f}")
print("  Acquisition Strategy: " + generated_plan['marketing']['customer_acquisition_strategy'])
print("")
print("[SECTION 4: FINANCIALS & RISK MITIGATION]")
print("  Recommended Investment: LKR " + f"{generated_plan['financials']['recommended_strategy_investment_lkr']:,}")
print("  Monthly Operating Cost: LKR " + f"{generated_plan['financials']['recommended_monthly_operating_cost_lkr']:,}")
print("  Expected Monthly Revenue: LKR " + f"{generated_plan['financials']['expected_monthly_revenue_lkr']:,}")
print("  Projected Monthly Margin: LKR " + f"{generated_plan['financials']['projected_net_monthly_margin_lkr']:,}")
print("  SHAP Risk Countermeasure: " + generated_plan['financials']['primary_risk_mitigation'])
print("")
print("[SECTION 5: PHASED ROADMAP]")
print("  0-3 Months: " + generated_plan['roadmap']['phase_1_short_term_0_to_3_months'][0])
print("  3-12 Months: " + generated_plan['roadmap']['phase_2_medium_term_3_to_12_months'][0])
print("  1+ Years: " + generated_plan['roadmap']['phase_3_long_term_1_year_plus'][0])
print("==================================================")
print("")
```

## 4. Assemble & Export Structured Business Profile
Constructing the final standardized machine-readable **Structured Business Profile (JSON)** for Component 1.

### Integration Schema Design:
- **`Component_1`**: Output payload
- **Consumable by**:
  - **Component 2**: Legal, Regulatory & Ethical Requirement Identification
  - **Component 3**: Financial Monitoring & Fraud/Anomaly Detection
  - **Component 4**: GNN Business Risk Prediction & Scalable Growth


```python
structured_business_profile = {
    "metadata": {
        "component_id": "Component_1",
        "component_name": "AI-Based Business Feasibility Analysis & Personalized Growth Plan Recommendation",
        "profile_schema_version": "1.0",
        "case_id": topsis_payload["case_id"],
        "target_region": "Colombo District, Sri Lanka"
    },
    "business": {
        "business_stage": raw_input.get("business_stage"),
        "business_category": raw_input.get("business_category"),
        "location_type": raw_input.get("location_type"),
        "district": raw_input.get("district"),
        "province": raw_input.get("province"),
        "proposed_action": raw_input.get("proposed_action")
    },
    "financial": {
        "available_capital_lkr": float(raw_input.get("available_capital_lkr", 0)),
        "loan_amount_lkr": float(raw_input.get("loan_amount_lkr", 0)),
        "monthly_budget_lkr": float(raw_input.get("monthly_budget_lkr", 0)),
        "initial_inventory_cost_lkr": float(raw_input.get("initial_inventory_cost_lkr", 0)),
        "expected_price_lkr": float(raw_input.get("expected_price_lkr", 0))
    },
    "market": {
        "expected_customers_per_day": int(raw_input.get("expected_customers_per_day", 0)),
        "customer_demand_score": float(raw_input.get("customer_demand_score", 0)),
        "competition_level": raw_input.get("competition_level"),
        "expected_operating_days_per_month": int(raw_input.get("expected_operating_days_per_month", 26))
    },
    "resources": {
        "entrepreneur_experience_years": float(raw_input.get("entrepreneur_experience_years", 0)),
        "location_suitability_score": float(raw_input.get("location_suitability_score", 0)),
        "available_staff_count": int(raw_input.get("available_staff_count", 0)),
        "required_staff_count": int(raw_input.get("required_staff_count", 0)),
        "available_equipment_score": float(raw_input.get("available_equipment_score", 0)),
        "required_equipment_score": float(raw_input.get("required_equipment_score", 0)),
        "supplier_availability_score": float(raw_input.get("supplier_availability_score", 0))
    },
    "feasibility": {
        "predicted_class": pred_info["predicted_class"],
        "probabilities": pred_info["probabilities"]
    },
    "explanation": {
        "shap_version": shap_payload.get("shap_version"),
        "expected_values": shap_payload.get("expected_values"),
        "top_global_features": shap_payload.get("top_global_features")[:5]
    },
    "strategies": topsis_payload["all_ranked_strategies"],
    "selected_strategy": topsis_payload["top_ranked_strategy"],
    "what_if_analysis": {
        "simulated_scenarios": what_if_payload.get("simulated_scenarios"),
        "counterfactual_capital_search": what_if_payload.get("counterfactual_capital_search")
    },
    "personalized_business_plan": generated_plan
}

profile_export_path = "../models/feasibility_model/structured_business_profile.json"
with open(profile_export_path, "w", encoding="utf-8") as f:
    json.dump(structured_business_profile, f, indent=4)

print(f"Successfully exported Structured Business Profile to '{profile_export_path}'")
```
